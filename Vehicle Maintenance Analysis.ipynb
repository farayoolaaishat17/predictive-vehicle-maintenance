### Frequency & Percentage Analysis

```python
def freq_percent(col):
    freq = data[col].value_counts()
    percent = data[col].value_counts(normalize = True) * 100
    table = pd.DataFrame({'Frequency': freq,
                         'Percentage (%)': percent.round(2)})
    return table
```

```python
print('=== Need Maintenance vs Mileage / Vehicle Age / Fuel Efficiency Groupby Analysis ===')
data.groupby('Need_Maintenance')[['Mileage', 'Vehicle_Age', 'Fuel_Efficiency']].mean().round(2)
```

### Pivot Table Analysis

```python
# List of categorical variables for pivot analysis
categorical_vars = ['Maintenance_History', 'Vehicle_Model', 'Owner_Type', 'Fuel_Type', 'Transmission_Type',
                   'Tire_Condition', 'Brake_Condition', 'Battery_Status', 'Accident_History']
# Loop through each categorical variable and create pivot table
for col in categorical_vars:
    pivot = pd.pivot_table(data,
                          values = 'Need_Maintenance', # target variable
                          index = col, # row category
                          aggfunc = 'mean' # mean = % of vehicles needing maintenance
                          ) * 100
    print(f'\nPivot Table for {col} vs Need Maintenance (%): \n')
    print(pivot.round(2))
```

### Cross Tabulation Analysis

```python
# List of categorical variables for cross tabulation analysis
categorical_vars = ['Maintenance_History', 'Vehicle_Model', 'Owner_Type', 'Fuel_Type', 'Transmission_Type',
                   'Tire_Condition', 'Brake_Condition', 'Battery_Status', 'Accident_History']
# Loop through each categorical variable and create crosstab
for col in categorical_vars:
    cross_pct = pd.crosstab(data[col], data['Need_Maintenance'], normalize = 'index') * 100
    print(f'\nCross-tab for {col} vs Need Maintenance (%): \n')
    print(cross_pct.round(2))
```

### Pie Charts

```python
import matplotlib.pyplot as plt

pie_vars = ['Need_Maintenance', 'Maintenance_History', 'Tire_Condition', 'Brake_Condition', 'Battery_Status']

colors_list = [
    ['#99ff99', '#ff6666'], # need_maintenance
    ['#66b3ff', '#99ff99', '#ff9999'], # maintenance_history
    ['#66b3ff', '#ffcc99', '#ff9999'], # tire_condition
    ['#66b3ff', '#ffcc99', '#ff9999'], # brake_condition
    ['#66b3ff', '#ffcc99', '#ff9999'] # battery_status
]

for i, col in enumerate(pie_vars):
    counts = data[col].value_counts()
    plt.figure(figsize = (5, 5))
    plt.pie(counts, labels = counts.index, autopct = '%1.1f%%', startangle = 90, colors = colors_list[i])
    plt.title(f'Distribution of {col}')
    plt.show()
```

### Clustered Bar Charts

```python
import seaborn as sns
import matplotlib.pyplot as plt

condition_vars = ['Maintenance_History', 'Tire_Condition', 'Brake_Condition', 'Battery_Status']
for col in condition_vars:
    plt.figure(figsize = (6, 5))
    sns.countplot(data = data, x = col, hue = 'Need_Maintenance')
    plt.title(f'{col.replace('_', ' ').title()} vs Maintenance Requirement')
    plt.xlabel(col.replace('_', ' ').title())
    plt.ylabel('Number of Vehicles')
    plt.legend(title = 'Need_Maintenance', labels = ['No', 'Yes'])
    plt.show()
```

### Histogram

```python
import seaborn as sns
import matplotlib.pyplot as plt

numeric_vars = ['Reported_Issues', 'Mileage', 'Odometer_Reading', 'Vehicle_Age']
for col in numeric_vars:
    plt.figure(figsize = (6, 5))
    sns.histplot(data[col], bins = 20, kde = True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

### Kde Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize = (6, 5))
sns.kdeplot(data = data, x = 'Reported_Issues', hue = 'Need_Maintenance', fill = True, common_norm = False)
plt.title('KDE of Reported Issues by Maintenance Requirement')
plt.xlabel('Reported Issues')
plt.ylabel('Density')
plt.show()
```
